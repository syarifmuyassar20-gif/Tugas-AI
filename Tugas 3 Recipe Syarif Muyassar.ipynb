## 1. Recipe Audit Station CSV

### Problem

Baca satu file CSV stasiun dan identifikasi masalah pada skema kolom,
rentang waktu, nilai hilang, serta timestamp duplikat. Hasilnya berupa
ringkasan QA/QC, bukan analisis iklim atau prediksi hujan. File sumber
tidak boleh diubah.

### Minimal Theory

Audit data dilakukan sebelum agregasi atau analisis agar kesalahan
struktur, waktu, dan nilai dapat diketahui lebih awal. Timestamp ganda
dapat memengaruhi agregasi, sedangkan nilai hilang atau tipe kolom yang
tidak sesuai dapat mengubah hasil analisis.

### Input dan Constraints

-   `CSV_PATH`: path file CSV.
-   `TIME_COLUMN`: nama kolom waktu.
-   `UNITS`: kontrak satuan untuk kolom yang relevan.
-   Pertahankan file sumber; gunakan pembacaan *read-only*.
-   Jangan menghapus atau memperbaiki baris sumber secara otomatis.
-   Jangan menyimpulkan hasil ilmiah yang tidak diuji oleh audit.

### Prompt dan Procedure

> Buat draft recipe untuk audit QA/QC station CSV. Periksa skema kolom,
> rentang waktu, nilai hilang, timestamp duplikat, dan tipe data tanpa
> mengubah input mentah. Nyatakan input, teori minimum, asumsi,
> prosedur, output yang diharapkan, rencana implementasi Python,
> pemeriksaan kegagalan, serta status eksekusi. Pisahkan spesifikasi
> recipe dari klaim bahwa program telah dijalankan.

Langkah kerja: 1. Baca CSV secara *read-only*. 2. Periksa apakah kolom
wajib tersedia. 3. Validasi tipe data dan rentang timestamp. 4. Hitung
nilai hilang dan timestamp duplikat. 5. Buat laporan QA/QC dan catatan
pemeriksaan. 6. Catat error atau kolom yang tidak sesuai tanpa mengubah
sumber.

### Code / Implementation Plan

Implementasi dapat menggunakan Python dengan `pandas` untuk membaca CSV
dan menghitung ringkasan. Kode disimpan pada path relatif, misalnya
`code/recipes/io/r-io-001.py`. Path ini merupakan rancangan, bukan bukti
bahwa file sudah dibuat.

### Output

-   `results/io/station-summary.md`: laporan Markdown.
-   `results/run-records/r-io-001.json`: run record jika program
    dijalankan.

Laporan mencakup jumlah baris dan kolom, kolom yang ditemukan, rentang
waktu, jumlah nilai hilang, jumlah timestamp duplikat, dan status setiap
pemeriksaan.

### Optional Check / Evidence

Periksa bahwa file input tidak berubah, laporan tidak kosong, jumlah
baris sesuai dengan sumber, dan status eksekusi dicatat. Gunakan
`Not yet executed` jika program belum dijalankan. Label `Executed` hanya
digunakan jika bukti eksekusi tersedia.

## 2. Recipe Normalisasi GeoJSON

### Problem

Susun recipe untuk mengubah file GeoJSON menjadi GeoJSON yang
dinormalisasi, inventaris CSV, dan ringkasan JSON. Recipe harus
memeriksa geometri, properti, geometri kosong, serta urutan koordinat
tanpa menebak informasi yang tidak tersedia.

### Minimal Theory

GeoJSON menggunakan struktur JSON untuk menyimpan fitur geografis.
Pemeriksaan perlu mencakup struktur `FeatureCollection`, tipe geometri,
koordinat, properti, dan sistem referensi koordinat. Untuk koordinat
geografis GeoJSON, urutan koordinat umumnya berupa bujur lalu lintang.

### Input dan Constraints

-   `GEOJSON_PATH`: path GeoJSON sumber.
-   Path output untuk GeoJSON, CSV inventaris, dan JSON ringkasan.
-   Jangan menghilangkan fitur atau properti tanpa aturan yang jelas.
-   Jangan menebak CRS atau mengubah urutan koordinat tanpa dasar.
-   Keberhasilan membaca file tidak otomatis membuktikan topologi
    geometri valid.

### Prompt dan Procedure

> Susun recipe untuk menormalisasi `[GEOJSON_PATH]` menjadi GeoJSON,
> inventaris CSV, dan ringkasan JSON. Periksa struktur, tipe geometri,
> properti, geometri kosong atau tidak valid, urutan koordinat, dan
> informasi CRS. Nyatakan ketidakpastian secara terbuka dan jangan
> mengklaim output telah dihasilkan sebelum eksekusi dibuktikan.

Langkah kerja: 1. Baca dan validasi sintaks JSON. 2. Periksa struktur
objek dan daftar fitur. 3. Periksa geometri, koordinat, serta properti.
4. Identifikasi CRS atau urutan koordinat yang belum jelas. 5.
Normalisasi hanya berdasarkan aturan yang terdefinisi. 6. Ekspor
inventaris fitur dan ringkasan hasil pemeriksaan. 7. Catat fitur yang
gagal diproses.

### Code / Implementation Plan

Implementasi dapat menggunakan Python dengan pustaka yang mendukung
GeoJSON, misalnya `json` untuk pemeriksaan struktur dan pustaka
geospasial yang tersedia di lingkungan kerja. Dependency harus dicatat
dan tidak boleh diasumsikan sudah terpasang.

### Output

-   GeoJSON hasil normalisasi.
-   CSV inventaris fitur.
-   JSON ringkasan QA/QC.
-   Run record serta peringatan jika tersedia.

### Optional Check / Evidence

Bandingkan jumlah fitur sumber dan hasil, periksa properti penting,
validasi struktur output, dan catat fitur yang tidak berhasil diproses.
Jika program belum dijalankan, sebutkan output sebagai hasil yang
direncanakan.

## 3. Recipe Model Difusi Panas Satu Dimensi

### Problem

Buat recipe untuk mensimulasikan perubahan suhu terhadap posisi dan
waktu berdasarkan profil awal, difusivitas termal, kondisi batas, jarak
grid, dan langkah waktu.

### Minimal Theory

Persamaan difusi panas satu dimensi adalah:

\[
\frac{\partial T}{\partial t}=\alpha\frac{\partial^2T}{\partial x^2}
\]

dengan (T) adalah suhu, (t) waktu, (x) posisi, dan (\alpha)
difusivitas termal.

Untuk skema beda hingga eksplisit standar, parameter kestabilan dapat
ditulis:

\[ r=\frac{\alpha\Delta t}{(\Delta x)^2} \]

Pada kasus satu dimensi dengan skema eksplisit standar, syarat
kestabilan yang umum digunakan adalah (r \le 1/2). Syarat ini
perlu disesuaikan dengan skema numerik yang benar-benar digunakan.

### Input dan Constraints

-   `INITIAL_PROFILE`: profil suhu awal.
-   `ALPHA`: difusivitas termal dan satuannya.
-   `BOUNDARIES`: kondisi batas.
-   `DZ`: jarak grid.
-   `DT`: langkah waktu.
-   Pastikan konsistensi satuan.
-   Jangan mengklaim kalibrasi atau konvergensi tanpa pengujian.

### Prompt dan Procedure

> Buat draft recipe model difusi panas satu dimensi menggunakan profil
> awal `[INITIAL_PROFILE]`, difusivitas `[ALPHA]`, kondisi batas
> `[BOUNDARIES]`, jarak grid `[DZ]`, dan langkah waktu `[DT]`. Sertakan
> persamaan pengatur, konvensi satuan dan tanda, pemeriksaan kestabilan,
> pembanding analitik atau kasus dasar, profil output, grafik, run
> record, serta peringatan jika parameter atau dependency belum
> tersedia. Jangan menyatakan eksekusi tanpa bukti.

Langkah kerja: 1. Validasi parameter dan satuan. 2. Bentuk grid ruang
dan waktu. 3. Terapkan kondisi awal serta kondisi batas. 4. Periksa
kestabilan skema. 5. Jalankan iterasi jika implementasi tersedia. 6.
Bandingkan hasil dengan kasus dasar atau solusi analitik jika
memungkinkan. 7. Simpan profil suhu, grafik, dan catatan eksekusi.

### Code / Implementation Plan

Implementasi dapat menggunakan Python, `numpy` untuk komputasi numerik,
dan `matplotlib` untuk visualisasi. Pilihan metode numerik, parameter,
serta dependency harus dicatat.

### Output

-   Profil suhu terhadap posisi dan waktu.
-   Grafik hasil simulasi.
-   Ringkasan parameter.
-   Run record berisi status dan peringatan.

### Optional Check / Evidence

Periksa kestabilan numerik, kondisi batas, konsistensi satuan, dan
kesesuaian dengan kasus dasar. Jangan mengklaim konvergensi, kalibrasi,
atau keberhasilan eksekusi tanpa bukti yang sesuai.

# BAGIAN II --- RECIPE BARU BUATAN SAYA

## 4. Recipe QA/QC dan Analisis Data Seismik Refraksi

### Problem

Menyusun workflow untuk memeriksa kualitas data waktu tempuh gelombang seismik refraksi dan membuat grafik waktu tempuh (*travel-time curve*) terhadap jarak geofon. Recipe ini membantu mengidentifikasi data yang hilang, nilai waktu tempuh yang tidak masuk akal, geometri sumber-geofon yang tidak konsisten, serta titik pengamatan yang berpotensi menjadi pencilan sebelum dilakukan estimasi kecepatan lapisan bawah permukaan.

### Minimal Theory

Metode seismik refraksi memanfaatkan waktu tempuh gelombang dari sumber menuju geofon. Pada model dua lapisan horizontal dengan lapisan kedua memiliki kecepatan lebih tinggi, hubungan waktu tempuh gelombang langsung dapat ditulis:

\[
t_d = \frac{x}{v_1}
\]

Untuk gelombang refraksi kritis, bentuk sederhana waktu tempuhnya adalah:

\[
t_r = t_i + \frac{x}{v_2}
\]

dengan \(x\) adalah jarak sumber-geofon, \(v_1\) adalah kecepatan lapisan atas, \(v_2\) adalah kecepatan lapisan bawah, dan \(t_i\) adalah waktu intersep. Persamaan tersebut merupakan pendekatan model dua lapisan sederhana; interpretasi aktual harus mempertimbangkan geometri survei dan asumsi model.

Kecepatan dapat diperkirakan dari kebalikan kemiringan garis pada grafik waktu tempuh terhadap jarak. Perubahan kemiringan dapat menunjukkan perubahan kecepatan rambat, tetapi tidak secara otomatis membuktikan jumlah atau jenis lapisan geologi.

### Input dan Constraints

Input utama:
- `DATA_PATH`: lokasi file CSV data seismik refraksi.
- `SHOT_ID`: identitas tembakan/sumber.
- `GEOPHONE_ID`: identitas geofon.
- `OFFSET_M`: jarak sumber ke geofon dalam meter.
- `FIRST_ARRIVAL_MS`: waktu tiba pertama dalam milidetik.

Input tambahan jika tersedia:
- `LINE_ID`: identitas lintasan survei.
- `SHOT_COORDINATE`: posisi sumber.
- `GEOPHONE_COORDINATE`: posisi geofon.
- `PICK_QUALITY`: penilaian kualitas picking atau tingkat keyakinan.

Batasan:
1. File sumber dibaca secara read-only dan tidak ditimpa.
2. Jarak harus menggunakan satuan meter dan waktu harus memiliki satuan yang konsisten.
3. Nilai jarak atau waktu yang hilang, negatif, atau tidak valid harus ditandai, bukan diperbaiki diam-diam.
4. Data dari tembakan berbeda tidak boleh digabungkan tanpa memeriksa geometri dan identitasnya.
5. Pencilan tidak boleh dihapus otomatis tanpa aturan yang terdokumentasi.
6. Kecepatan hasil fitting merupakan estimasi awal dan tidak boleh langsung dianggap sebagai identifikasi litologi.

### Prompt dan Procedure

**Prompt yang dapat digunakan kembali:**

> Anda bertindak sebagai asisten pengolahan data seismik refraksi. Baca `[DATA_PATH]` secara read-only dan periksa kolom wajib `SHOT_ID`, `GEOPHONE_ID`, `OFFSET_M`, dan `FIRST_ARRIVAL_MS`. Identifikasi nilai hilang, nilai negatif, duplikasi pasangan tembakan-geofon, ketidakkonsistenan satuan, dan geometri yang perlu diperiksa. Buat grafik waktu tiba pertama terhadap offset, bedakan data berdasarkan tembakan jika diperlukan, dan lakukan fitting garis pada segmen yang secara visual dan fisik layak. Estimasikan kecepatan dari kemiringan garis serta waktu intersep jika modelnya sesuai. Laporkan asumsi, ketidakpastian, residu, dan keterbatasan. Jangan mengklaim jumlah lapisan atau jenis batuan hanya dari grafik. Jika program belum dijalankan, tulis `Not yet executed`.

**Procedure:**
1. Baca file CSV tanpa mengubah data mentah.
2. Periksa kolom wajib, tipe data, satuan, dan jumlah baris.
3. Tandai nilai kosong, offset negatif, waktu tiba tidak valid, serta pasangan tembakan-geofon duplikat.
4. Pisahkan data berdasarkan lintasan atau identitas tembakan.
5. Buat grafik *travel-time curve* dengan offset pada sumbu-x dan waktu tiba pada sumbu-y.
6. Periksa apakah terdapat segmen linear yang masuk akal berdasarkan data dan asumsi model.
7. Lakukan regresi linear pada segmen yang dipilih dengan mencatat alasan pemilihannya.
8. Estimasikan kecepatan dari kebalikan kemiringan garis; untuk model refraksi sederhana, catat waktu intersep bila relevan.
9. Hitung residu dan metrik kesalahan fitting, lalu tandai titik dengan residu besar untuk ditinjau.
10. Simpan grafik, tabel hasil, laporan QA/QC, dan status eksekusi.

### Code / Implementation Plan

Implementasi dapat menggunakan Python dengan:
- `pandas` untuk membaca data dan melakukan audit tabel.
- `numpy` untuk operasi numerik.
- `matplotlib` untuk grafik waktu tempuh dan pemeriksaan visual.
- `scipy.stats` atau `numpy.polyfit` untuk regresi linear pada segmen yang dipilih.
- `json` dan `pathlib` untuk metadata serta pencatatan hasil.

Contoh struktur direktori proyek:

```text
seismic-refraction-qc/
├── data/
│   └── first-arrivals.csv
├── code/
│   └── refraction_travel_time_qc.py
└── results/
    ├── qc-report.md
    ├── travel-time-curve.png
    ├── fitted-segments.csv
    └── run-record.json
```

Struktur tersebut adalah rancangan direktori, bukan bukti bahwa file output telah dibuat.

### Output

1. `qc-report.md`: ringkasan jumlah data, kolom yang tersedia, data tidak valid, duplikasi, serta catatan QA/QC.
2. `travel-time-curve.png`: grafik waktu tiba pertama terhadap offset, dibedakan berdasarkan tembakan bila diperlukan.
3. `fitted-segments.csv`: tabel segmen fitting, kemiringan, estimasi kecepatan, waktu intersep, jumlah titik, dan metrik residu.
4. `run-record.json`: waktu eksekusi, parameter, versi pustaka jika tersedia, peringatan, dan status proses.

### Optional Check / Evidence

- [ ] Data mentah tidak berubah setelah proses.
- [ ] Kolom wajib dan satuan sudah diperiksa.
- [ ] Offset dan waktu tiba yang tidak valid telah ditandai.
- [ ] Identitas tembakan dan geofon dipertahankan.
- [ ] Grafik menggunakan satuan yang jelas.
- [ ] Pemilihan segmen regresi dijelaskan dan dapat diulang.
- [ ] Estimasi kecepatan dihitung dari kemiringan dengan konversi satuan yang benar.
- [ ] Residu dan keterbatasan fitting dilaporkan.
- [ ] Status eksekusi sesuai bukti yang tersedia.

Status yang diperbolehkan:
- `Draft`: spesifikasi recipe sedang disusun.
- `Not yet executed`: program belum dijalankan.
- `Partial`: QA/QC selesai, tetapi fitting atau ekspor belum selesai.
- `Executed`: program telah dijalankan dan output yang dilaporkan benar-benar tersedia.

### Keterbatasan dan Risiko

Interpretasi seismik refraksi bergantung pada kualitas picking, geometri akuisisi, kontras kecepatan, dan asumsi model bawah permukaan. Lapisan berkecepatan rendah di bawah lapisan berkecepatan lebih tinggi dapat tidak terdeteksi dengan baik oleh refraksi konvensional. Grafik waktu tempuh yang tampak memiliki beberapa segmen linear juga tidak selalu berarti jumlah lapisan sama dengan jumlah segmen. Hasil harus diperlakukan sebagai estimasi awal dan, jika memungkinkan, dibandingkan dengan informasi geologi atau metode geofisika lain.

### Ringkasan Recipe Baru

Recipe ini menyediakan alur kerja QA/QC untuk data waktu tiba pertama seismik refraksi, mulai dari pemeriksaan struktur data dan geometri, visualisasi kurva waktu tempuh, hingga estimasi kecepatan melalui regresi pada segmen yang sesuai. Setiap asumsi, peringatan, dan status eksekusi dicatat agar hasil dapat diperiksa ulang dan tidak digunakan untuk menyimpulkan kondisi geologi secara berlebihan.
