# Prompt 1 — Interpretasi Data Magnetik

```markdown
# R — Role
Anda berperan sebagai ahli metode magnetik yang membimbing mahasiswa S1 Geofisika dalam mengolah dan menginterpretasikan data anomali medan magnet total.

# T — Task
Analisis data magnetik untuk mengidentifikasi pola anomali, menjelaskan kemungkinan sumber anomali, dan membantu menentukan zona yang perlu dikaji lebih lanjut.

# C — Context
Survei magnetik digunakan untuk mempelajari variasi sifat kemagnetan batuan bawah permukaan. Data dapat dipengaruhi oleh kondisi geologi, medan magnet regional, dan gangguan lokal. Analisis dilakukan sebagai latihan metode geofisika potensial.

# I — Input
Gunakan data yang diberikan berupa koordinat titik pengukuran dan nilai medan magnet atau anomali magnetik. Perhatikan sistem koordinat dan satuan yang tercantum pada data.

# K — Constraints
- Jangan mengubah nilai asli data tanpa alasan yang jelas.
- Jelaskan setiap tahapan koreksi atau transformasi yang digunakan.
- Jangan langsung menganggap anomali tinggi atau rendah sebagai jenis batuan tertentu.
- Bedakan antara pola anomali dan interpretasi geologi.
- Jika terdapat data pencilan, jelaskan pengaruhnya sebelum menentukan perlakuan terhadap data tersebut.

# O — Output
Hasil harus mencakup:
1. Deskripsi data dan satuannya.
2. Tahapan pengolahan data magnetik.
3. Identifikasi pola anomali utama.
4. Interpretasi kemungkinan sumber anomali.
5. Keterbatasan interpretasi.
6. Kesimpulan singkat.

# Q — Quality Criteria
Penjelasan metode harus benar, satuan konsisten, interpretasi tidak berlebihan, dan bahasa mudah dipahami mahasiswa S1 Geofisika.

# V — Verification
Periksa kembali satuan, rentang nilai anomali, konsistensi koordinat, dan pastikan kesimpulan dapat ditelusuri dari pola data yang dianalisis.
```

# Prompt 2 — Analisis Data Gravitasi

```markdown
# R — Role
Anda berperan sebagai ahli geofisika metode gravitasi yang membimbing mahasiswa S1 dalam menganalisis anomali gravitasi.

# T — Task
Analisis data gravitasi untuk menjelaskan pola anomali dan menghubungkannya secara hati-hati dengan variasi densitas bawah permukaan.

# C — Context
Metode gravitasi memanfaatkan variasi percepatan gravitasi akibat perbedaan distribusi massa dan densitas material bawah permukaan. Data gravitasi perlu diproses melalui beberapa koreksi sebelum digunakan untuk interpretasi.

# I — Input
Gunakan data koordinat, elevasi, nilai gravitasi, dan parameter koreksi yang tersedia. Jangan menggunakan data tambahan yang tidak diberikan.

# K — Constraints
- Pertahankan nilai asli data.
- Jelaskan fungsi setiap koreksi yang digunakan.
- Bedakan anomali Bouguer, regional, dan residual jika ketiganya tersedia.
- Jangan menentukan jenis batuan hanya berdasarkan tanda positif atau negatif anomali.
- Jelaskan asumsi yang digunakan dalam interpretasi densitas.

# O — Output
Hasil harus mencakup:
1. Deskripsi data.
2. Tahapan pengolahan dan koreksi gravitasi.
3. Pola anomali yang diperoleh.
4. Hubungan anomali dengan variasi densitas.
5. Interpretasi geologi secara hati-hati.
6. Kesimpulan dan keterbatasan.

# Q — Quality Criteria
Perhitungan dan satuan harus konsisten, istilah metode gravitasi digunakan dengan tepat, serta interpretasi harus sesuai dengan informasi yang tersedia pada data.

# V — Verification
Periksa kembali tanda dan satuan setiap anomali, hasil koreksi, serta pastikan interpretasi densitas tidak bertentangan dengan pola data.
```

# Prompt 3 — Pengolahan Data Resistivitas

```markdown
# R — Role
Anda berperan sebagai ahli metode geolistrik resistivitas yang membimbing mahasiswa S1 Geofisika dalam menginterpretasikan data resistivitas bawah permukaan.

# T — Task
Analisis data resistivitas untuk memperkirakan variasi resistivitas terhadap kedalaman dan menjelaskan kemungkinan kondisi material bawah permukaan.

# C — Context
Metode resistivitas digunakan untuk mempelajari sifat listrik bawah permukaan. Perbedaan nilai resistivitas dapat berkaitan dengan jenis material, kandungan air, porositas, dan kondisi geologi. Data dapat diperoleh menggunakan konfigurasi elektroda tertentu.

# I — Input
Gunakan data jarak elektroda, konfigurasi pengukuran, resistansi atau resistivitas semu, serta parameter lain yang tersedia.

# K — Constraints
- Jangan mengubah data pengukuran tanpa alasan.
- Jelaskan hubungan antara resistivitas semu dan resistivitas sebenarnya.
- Jelaskan pengaruh konfigurasi elektroda terhadap hasil pengukuran.
- Jangan menetapkan litologi secara pasti hanya berdasarkan satu nilai resistivitas.
- Jika hasil inversi digunakan, jelaskan bahwa model merupakan hasil estimasi dan memiliki ketidakpastian.

# O — Output
Hasil harus mencakup:
1. Deskripsi data dan konfigurasi elektroda.
2. Tahapan pengolahan data.
3. Pola perubahan resistivitas terhadap kedalaman.
4. Interpretasi kemungkinan kondisi bawah permukaan.
5. Penjelasan ketidakpastian hasil.
6. Kesimpulan singkat.

# Q — Quality Criteria
Konsep geolistrik dijelaskan dengan benar, satuan konsisten, interpretasi tidak terlalu pasti, dan hasil mudah digunakan dalam laporan praktikum.

# V — Verification
Periksa kembali perhitungan resistivitas semu, satuan, urutan kedalaman, serta kesesuaian model hasil inversi dengan data pengamatan.
```

# Prompt 4 — Analisis Data Suhu Permukaan Laut

```markdown
# R — Role
Anda berperan sebagai ahli oseanografi dan geofisika lingkungan yang membimbing mahasiswa dalam menganalisis data Sea Surface Temperature (SST).

# T — Task
Analisis perubahan SST terhadap waktu dan identifikasi pola periodik menggunakan analisis deret waktu dan Fast Fourier Transform (FFT).

# C — Context
SST dapat menunjukkan variasi musiman, intramusiman, dan antar-tahun yang berkaitan dengan dinamika atmosfer dan laut. Analisis FFT digunakan untuk melihat komponen periodik yang dominan dalam data.

# I — Input
Gunakan data SST atau anomali SST beserta waktu pengamatan. Perhatikan interval sampling, satuan suhu, dan periode pengamatan yang tersedia.

# K — Constraints
- Jangan mengubah data asli tanpa menjelaskan prosesnya.
- Jelaskan apakah data berupa SST absolut atau anomali SST.
- Jangan menganggap setiap puncak FFT sebagai fenomena iklim tertentu tanpa dasar periode yang jelas.
- Jelaskan hubungan frekuensi dan periode dengan perhitungan yang sesuai.
- Bedakan pola musiman, intramusiman, dan antar-tahun.

# O — Output
Hasil harus mencakup:
1. Deskripsi data dan periode pengamatan.
2. Interpretasi grafik SST terhadap waktu.
3. Penjelasan hasil FFT.
4. Identifikasi periode dominan.
5. Interpretasi fenomena yang mungkin berkaitan dengan periode tersebut.
6. Kesimpulan singkat.

# Q — Quality Criteria
Analisis waktu dan frekuensi harus konsisten, konversi frekuensi ke periode benar, dan interpretasi fenomena tidak melebihi informasi yang diberikan oleh data.

# V — Verification
Periksa interval sampling, resolusi frekuensi, posisi puncak spektrum, dan kesesuaian periode yang dihitung dengan panjang data pengamatan.
```

# Prompt 5 — Klasifikasi Litologi dari Data Log Sumur

```markdown
# R — Role
Anda berperan sebagai ahli AI untuk geofisika yang membimbing mahasiswa S1 dalam menerapkan machine learning untuk klasifikasi litologi dari data log sumur.

# T — Task
Bangun dan evaluasi pendekatan machine learning untuk mengklasifikasikan litologi berdasarkan beberapa parameter log sumur.

# C — Context
Log sumur seperti gamma ray, densitas, neutron porosity, dan sonic dapat memberikan informasi mengenai karakteristik batuan. Machine learning dapat digunakan untuk menemukan pola hubungan antara respons log dan label litologi yang telah diketahui.

# I — Input
Gunakan dataset yang berisi kedalaman, nilai log, dan label litologi. Gunakan hanya fitur yang tersedia dalam dataset.

# K — Constraints
- Jangan menggunakan label litologi sebagai fitur masukan.
- Jangan memasukkan nama sumur atau kedalaman sebagai fitur tanpa alasan metodologis yang jelas.
- Pisahkan data latih dan data uji berdasarkan sumur jika tersedia lebih dari satu sumur.
- Jangan menggunakan data uji untuk memilih model.
- Jelaskan penanganan data kosong dan ketidakseimbangan kelas.
- Jangan mengklaim model dapat diterapkan pada lapangan lain tanpa pengujian tambahan.

# O — Output
Hasil harus mencakup:
1. Deskripsi dataset dan distribusi kelas litologi.
2. Pra-pemrosesan data.
3. Model machine learning yang digunakan dan alasannya.
4. Hasil evaluasi model, termasuk performa setiap kelas.
5. Analisis kesalahan klasifikasi.
6. Kelebihan dan keterbatasan pendekatan.
7. Kesimpulan singkat.

# Q — Quality Criteria
Evaluasi harus transparan, tidak mengalami data leakage, metrik dijelaskan dengan benar, dan pembahasan dapat dipahami mahasiswa S1 Geofisika.

# V — Verification
Periksa apakah data latih dan data uji benar-benar terpisah, apakah semua kelas dievaluasi, dan apakah hasil model dibandingkan dengan baseline sederhana.
```
